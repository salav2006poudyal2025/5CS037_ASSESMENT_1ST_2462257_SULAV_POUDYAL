```python
# Task-3. Outlier Detection:
outlier_df = df_south_asia[['country', 'hdi', 'gross_inc_percap']]
```

```python
# Define a function to detect outliers using the 1.5 × IQR rule
def detect_outliers(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return series[(series < lower_bound) | (series > upper_bound)]
```

```python
# Detect outliers in HDI
hdi_outliers = detect_outliers(outlier_df['hdi'])
hdi_outliers
```

```python
# Detect outliers in GNI per Capita
gni_outliers = detect_outliers(outlier_df['gross_inc_percap'])
gni_outliers
```

```python
# Create a scatter plot highlighting outliers
# Identify outlier countries
outlier_countries = outlier_df[
    (outlier_df['hdi'].isin(hdi_outliers)) |
    (outlier_df['gross_inc_percap'].isin(gni_outliers))
]
plt.figure(figsize=(6, 5))
plt.scatter(outlier_df['gross_inc_percap'], outlier_df['hdi'], label='Normal', alpha=0.6)
plt.scatter(
    outlier_countries['gross_inc_percap'],
    outlier_countries['hdi'],
    color='red',
    label='Outliers'
)

plt.xlabel("GNI per Capita")
plt.ylabel("HDI")
plt.title("Outliers in HDI vs GNI per Capita (South Asia)")
plt.legend()
plt.show()
```

```python
# Discuss why the identified countries stand out as outliers.
# For HDI no outliers were detected using the IQR-based method.
# This indicates that HDI values across the countries are relatively concentrated within a narrow range with no country showing an extremely high or low level of overall human development compared to the rest.
# HDI being a composite index bounded between 0 and 1 typically exhibits less extreme variation which explains the absence of detected outliers.

# For Gross National Income (GNI) per capita shows several clear outliers.
# The listed countries have substantially higher income levels compared to the majority of countries in the dataset causing them to fall outside the typical income range.
# Economic indicators such as GNI per capita tend to have wider dispersion because income is strongly influenced by factors like natural resource endowments, industrial structure, remittances and integration into global markets.

# The presence of GNI outliers highlights that income distribution across countries is far more uneven than overall human development. While some countries achieve exceptionally high income levels this does not always translate into equally extreme HDI values, reinforcing the idea that human development outcomes are more balanced than pure economic measures.
```

```python
# Plot the top 3 positive gaps and top 3 negative gaps.
import matplotlib.pyplot as plt
import seaborn as sns


top_positive = gap_desc.head(3).reset_index()  # 'country' and 'GNI_HDI_Gap'
top_negative = gap_asc.head(3).reset_index()

# Combine top positive and negative gaps
gap_plot_df = pd.concat([top_positive, top_negative], ignore_index=True)


plt.figure(figsize=(8, 5))
sns.barplot(
    data=gap_plot_df,
    x='GNI_HDI_Gap',
    y='country',
    orient='h'
)

plt.xlabel('GNI–HDI Gap')
plt.ylabel('Country')
plt.title('Top 3 Positive and Negative GNI–HDI Gaps (South Asia)')
plt.tight_layout()
plt.show()
```

```python
# Task-7. Outlier Detection:
# Detect outliers in HDI {"hdi"} and GNI per Capita {"gross inc percap"} for each region using the 1.5 × IQR rule.
# Create scatter plots highlighting outliers in a different color.
```

```python
def detect_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return (df[column] < lower) | (df[column] > upper)

for df in [df_south_asia, df_middle_east]:
    df['HDI_Outlier'] = detect_outliers_iqr(df, 'hdi')
    df['GNI_Outlier'] = detect_outliers_iqr(df, 'gross_inc_percap')
```

```python

# Create scatter plots for both regions side by side
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# South Asia
sa_normal = df_south_asia[~(df_south_asia['HDI_Outlier'] | df_south_asia['GNI_Outlier'])]
sa_outliers = df_south_asia[df_south_asia['HDI_Outlier'] | df_south_asia['GNI_Outlier']]

axes[0].scatter(
    sa_normal['gross_inc_percap'],
    sa_normal['hdi'],
    c='blue', label='Normal', alpha=0.6, s=50
)
axes[0].scatter(
    sa_outliers['gross_inc_percap'],
    sa_outliers['hdi'],
    c='red', label='Outliers', alpha=0.8, s=100, edgecolors='black', linewidth=1.5
)
axes[0].set_xlabel("GNI per Capita (PPP $)", fontsize=11)
axes[0].set_ylabel("HDI", fontsize=11)
axes[0].set_title("South Asia: HDI vs GNI (Outliers Highlighted)", fontsize=12, fontweight='bold')
axes[0].legend()
axes[0].grid(alpha=0.3)

# Middle East
me_normal = df_middle_east[~(df_middle_east['HDI_Outlier'] | df_middle_east['GNI_Outlier'])]
me_outliers = df_middle_east[df_middle_east['HDI_Outlier'] | df_middle_east['GNI_Outlier']]

axes[1].scatter(
    me_normal['gross_inc_percap'],
    me_normal['hdi'],
    c='blue', label='Normal', alpha=0.6, s=50
)
axes[1].scatter(
    me_outliers['gross_inc_percap'],
    me_outliers['hdi'],
    c='red', label='Outliers', alpha=0.8, s=100, edgecolors='black', linewidth=1.5
)
axes[1].set_xlabel("GNI per Capita (PPP $)", fontsize=11)
axes[1].set_ylabel("HDI", fontsize=11)
axes[1].set_title("Middle East: HDI vs GNI (Outliers Highlighted)", fontsize=12, fontweight='bold')
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.suptitle('Outlier Detection: HDI vs GNI per Capita (2020-2022)',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Discuss the significance of these outliers.
# Using the IQR method we have identified outliers in both HDI and GNI per capita for South Asia and the Middle East.

# 1. HDI Outliers
# HDI outliers are countries whose human development index is unusually high or low compared to the regional distribution.

# Significance:

# High HDI outliers:
# These countries perform exceptionally well in health, education, and income relative to neighbors.
# Example (South Asia): Maldives or Sri Lanka may appear as high outliers.

# Low HDI outliers:
# These indicate countries lagging in human development, possibly due to conflict, poor governance, or systemic poverty.
# Example (South Asia): Afghanistan is likely a low outlier.

# Implications:
# Highlight countries needing policy focus to improve quality of life.
# Can guide regional development support and targeted interventions.

# 2. GNI Per Capita Outliers
# GNI outliers reflect countries with unusually high or low income levels relative to their region.

# Significance:

# High GNI outliers:
# Often small or resource-rich economies with high per capita income.
# Example (Middle East): Qatar or UAE—high income but HDI may vary.

# Low GNI outliers:
# Indicate countries struggling economically.
# Example (South Asia): Afghanistan or Nepal.

# Implications:
# Reveals income inequality within regions.
# Highlights where high GNI does not translate to proportional human development (link to GNI–HDI gap analysis).

# 3. Overall Interpretation
# Outliers are critical for understanding extreme cases:
# They draw attention to success stories (high performers) and urgent development needs (low performers).
# Can inform policy decisions, aid allocation, and investment priorities.

# Combining HDI and GNI outliers can uncover mismatches:
# High GNI but low HDI: Indicates income isn’t effectively improving health and education.
# Low GNI but high HDI: Suggests efficient use of resources to improve human development despite low income.

# Summary Table:
# Outlier Type	High Outlier	Low Outlier	Significance
# HDI	Top-performing countries	Lagging countries	Highlights extreme development cases; guides policy
# GNI	Wealthy countries	Poor countries	Reveals income inequality and efficiency in converting income to HDI

# Key takeaway: Outliers help identify countries that need attention or can serve as models for development strategies.
```
